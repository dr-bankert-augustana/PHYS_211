#<font size = 6> <b> 4. Average Acceleration </b> </font>

---

##<font size = 5> <b>4.1 Average Acceleration Formula(s)</b> </font>

1. $\hspace{1 pc}
a_{\text{ave}} \eq
\dsfrac{v_{\text{f}} - v_{\text{i}}}{t_{\text{f}} - t_{\text{i}}}
$

##<font size = 5> <b>4.2 Average Acceleration Examples</b> </font>

###<font size = 4> <b>Example 1:</b></font>

An object which is travelling along a straight line is observed to have a velocity of 7 $\mps$ in the positive direction at time 1 $\seconds$. Later, at time 11 $\seconds$, the same object is measured to have a velocity of 12 $\mps$ in the positive direction. Determine the average acceleration of the object during the time interval between the measurements.

<br>

**Step 1: Get the Values from the Problem:**

<br>

$
\begin{array}{l l l}
\textrm{Velocities:} & v_{\text{i}} = 7 \mps\textrm{, } &
v_{\text{f}} = 12 \mps \\&&\\
\textrm{Times:} & t_{\text{i}} = 1  \seconds \textrm{, } & t_{\text{f}} = 11 \seconds
\end{array}
$

<br>

**Step 2: Get the Relevant Equation(s):**

<br>

$
a_{\text{ave}} \eq
\dsfrac{v_{\text{f}} - v_{\text{i}}}{t_{\text{f}} - t_{\text{i}}}
$

<br>

**Step 3: Plug In Values:**

<br>

$
\begin{array}{l l}
a_{\text{ave}}
& \eq \dsfrac{12 \mps - 7 \mps}{11 \seconds - 1 \seconds} \\&\\
& \eq \dsfrac{5 \mps}{10 \seconds} \\&\\
& \eq 0.5 \spc \mpss
\end{array}
$

<br>

---

<br>

###<font size = 4> <b>Example 2:</b></font>

A hiker is sliding backwards along a steep straight path with a velocity of $-3$ $\mps$ at time 2 $\seconds$. At time 9 $\seconds$, the hiker has managed to stop their slide and is traveling back up the path with a velocity of $2$ $\mps$. Determine the hiker's average acceleration during this time interval.

<br>

**Step 1: Get the Values from the Problem:**

<br>

$
\begin{array}{l l l}
\textrm{Velocities:} & v_{\text{i}} = -3 \mps\textrm{, } &
v_{\text{f}} = 2 \mps \\&&\\
\textrm{Times:} & t_{\text{i}} = 2 \seconds \textrm{, } & t_{\text{f}} = 9 \seconds
\end{array}
$

<br>

**Step 2: Get the Relevant Equation(s):**

<br>

$
a_{\text{ave}} \eq
\dsfrac{v_{\text{f}} - v_{\text{i}}}{t_{\text{f}} - t_{\text{i}}}
$

<br>

**Step 3: Plug In Values:**

<br>

$
\begin{array}{l l}
a_{\text{ave}}
& \eq \dsfrac{2 \mps - (-3) \mps}{9 \seconds - 2 \seconds} \\&\\
& \eq \dsfrac{5 \mps}{7 \seconds} \\&\\
& \eq 0.714 \spc \mpss
\end{array}
$

<br>

---

<br>

###<font size = 4> <b>Example 3:</b></font>

A car is traveling along a straight road with a velocity of 24 $\mps$ in the positive direction at time 5 $\seconds$. The car is has an average acceleration of 3 $\mpss$. What is the car's velocity in the positive direction at 13 $\seconds$?

<br>

**Step 1: Get the Values from the Problem:**

<br>

$
\begin{array}{l l l}
\textrm{Velocities:} & v_{\text{i}} = 24 \mps\textrm{, } &
v_{\text{f}} = \textrm{?} \\&&\\
\textrm{Times:} & t_{\text{i}} = 5 \seconds \textrm{, } & t_{\text{f}} = 13 \seconds \\&&\\
\textrm{Acceleration:} & a_{\textrm{ave}} = 3 \mpss & \\&&\\
\end{array}
$

<br>

**Step 2: Get the Relevant Equation(s):**

<br>

$
a_{\text{ave}} \eq
\dsfrac{v_{\text{f}} - v_{\text{i}}}{t_{\text{f}} - t_{\text{i}}}
$

<br>

**Step 3: Solve for the Desired Quantity:**

<br>

$
v_{\text{f}} \eq v_{\text{i}} + a_{\text{ave}} \spc \parens{t_{\text{f}} - t_{\text{i}}}
$

<br>

**Step 4: Plug In Values:**

<br>

$
\begin{array}{l l}
v_{\text{f}} & \eq v_{\text{i}} + a_{\text{ave}} \spc \parens{t_{\text{f}} - t_{\text{i}}} \\&\\
& \eq 24 \mps + 3 \mpss \spc \parens{13 \seconds - 5 \seconds} \\&\\
& \eq 24 \mps + 3 \mpss \spc \parens{8 \seconds} \\&\\
& \eq 24 \mps + 24 \mps \\&\\
& \eq 48 \mps
\end{array}
$
