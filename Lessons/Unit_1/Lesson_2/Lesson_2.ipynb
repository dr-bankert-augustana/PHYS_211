---

##<font size = 5> <b>1.1 Average Velocity Formula(s)</b> </font>

1. $\hspace{1 pc}
v_{\text{ave}} \eq
\dsfrac{x_{\text{f}} - x_{\text{i}}}{t_{\text{f}} - t_{\text{i}}}
$

<br>

---

<br>

##<font size = 5> <b>1.2 Average Velocity Example Problem(s)</b> </font>

###<font size = 4> <b>Example 1:</b></font>

A runner passes a marker located at 25 m along a straight track at t = 5s. At t = 17 s, the runner passes a marker located at 100 m along the straight track. What is the average velocity of the runner during this time interval?

<br>

**Step 1: Get the Values from the Problem:**

<br>

$
\begin{array}{l l l}
\textrm{Positions:} & x_{\text{i}} = 25 \textrm{ m} \textrm{, } &
x_{\text{f}} = 100 \textrm{ m} \\&&\\
\textrm{Times:} & t_{\text{i}} = 5  \seconds \textrm{, } & t_{\text{f}} = 17 \seconds
\end{array}
$

<br>

**Step 2: Get the Relevant Equation(s):**

<br>

$
v_{\text{ave}} \eq
\dsfrac{x_{\text{f}} - x_{\text{i}}}{t_{\text{f}} - t_{\text{i}}}
$

<br>

**Step 3: Plug In Values:**

<br>

$
\begin{array}{l l}
v_{\text{ave}}
& \eq \dsfrac{100 \textrm{ m} - 25 \textrm{ m}}{17 \seconds - 5 \seconds} \\&\\
& \eq \dsfrac{75 \textrm{ m}}{12 \seconds} \\&\\
& \eq 6.25 \spc \mps
\end{array}
$

##<font size = 5> <b>2.1 Motion with Constant Velocity Formula(s)</b> </font>

##<font size = 5> <b>2.2 Motion with Constant Velocity Example Problem(s)</b> </font>

###<font size = 4> <b>Example 1:</b></font>

A train moves along a straight section of track with a constant velocity of 15 m/s in the positive direction. At t = 10 s, the train is 320 m from a station. How far from the station will the train be at t = 25 s?

<br>

**Step 1: Get the Values from the Problem:**

<br>

$
\begin{array}{l l l}
\textrm{Positions:} & x_{\text{i}} = 320 \textrm{ m} \textrm{, } &
x_{\text{f}} = ??? \textrm{ m} \\&&\\
\textrm{Times:} & t_{\text{i}} = 10  \seconds \textrm{, } & t_{\text{f}} = 25 \seconds \\&\\
v_{\textrm{ave}} = 15 \mps
\end{array}
$

<br>

**Step 2: Get the Relevant Equation(s):**

<br>

$
x_{\text{f}} \eq
x_{\text{f}} \eq x_{\text{i}} + v_{\text{ave}} \spc \parens{t_{\text{f}} - t_{\text{i}}}
$

<br>

**Step 3: Plug In Values:**

<br>

$
\begin{array}{l l}
x_{\text{f}} & \eq 320 \textrm{ m} + 15 \mps \spc \parens{25 \textrm{ s} - \textrm{10 s}} \\&\\
& \eq 320 \textrm{ m} + 15 \mps \spc \parens{15 \textrm{ s}} \\&\\
& \eq 320 \textrm{ m} + 225 \textrm{ m} \\&\\
& \eq 545 \textrm{ m}
\end{array}
$

<br>

**Answer:**

<br>

The train will be **(545) m** from the station at (25) s.

#<font size = 6> <b> 3. Instantaneous Velocity </b> </font>

---

##<font size = 5> <b>3.1 Instantaneous Velocity Formula(s)</b> </font>

1. $\hspace{1 pc}
v\of{t} \eq \dsfrac{d \spc x\of{t}}{d \spc t}$

<br>

---

<br>

##<font size = 5> <b>3.2 Instantaneous Velocity Example Problem(s)</b> </font>

###<font size = 4> <b>Example 1:</b></font>

The position of a particle moving along the x-axis is described by:

<br>

$$x\of{t} = 3t^2 + 2t - 5$$

<br>

where x is measured in meters and t is measured in seconds. Determine the instantaneous velocity of the particle at t = 4.0 s.

<br>

**Step 1: Get the Values from the Problem:**

<br>

$
\begin{array}{l l l}
\textrm{Time:} & t = 4.0 \textrm{ s}
\end{array}
$

<br>

**Step 2: Get the Relevant Equation(s):**

<br>

$
v\of{t} \eq \dsfrac{d \spc x\of{t}}{d \spc t}
$

<br>

**Step 3: Evaluate the Expression:**

<br>

$
\begin{array}{l l}
v\of{t} & \eq \dsfrac{d \spc x\of{t}}{d \spc t} \\&\\
& \eq \dsfrac{d}{d \spc t} \spc \parens{3t^2 + 2t - 5} \\&\\
& \eq \dsfrac{d}{d \spc t} \spc \parens{3t^2} + \dsfrac{d}{d \spc t} \spc \parens{2t} - \dsfrac{d}{d \spc t} \spc \parens{5} \\&\\
& \eq 6t + 2 + 0 \\&\\
& \eq 6t + 2
\end{array}
$

<br>

**Step 4: Plug In Values:**

<br>

$
\begin{array}{l l}
v\of{t} & \eq 6 \mpss \spc \parens{4.0 \textrm{ s}} + 2 \mps \\&\\
& \eq 24 \mps + 2 \mps \\&\\
& \eq 26 \mps
\end{array}
$

##<font size = 5> <b>5.2 Motion with Constant Acceleration Example Problem(s)</b> </font>

A car is traveling along a straight road with a velocity of 24 $\mps$ in the positive direction at time 5 $\seconds$. The car is has a constant acceleration of 3 $\mpss$. What is the car's velocity in the positive direction at 13 $\seconds$?

<br>

**Step 1: Get the Values from the Problem:**

<br>

$
\begin{array}{l l l}
\textrm{Velocities:} & v_{\text{i}} = 24 \mps\textrm{, } &
v_{\text{f}} = \textrm{?} \\&&\\
\textrm{Times:} & t_{\text{i}} = 5 \seconds \textrm{, } & t_{\text{f}} = 13 \seconds \\&&\\
\textrm{Acceleration:} & a_{\textrm{ave}} = 3 \mpss & \\&&\\
\end{array}
$

<br>

**Step 2: Get the Relevant Equation(s):**

<br>

$
v_{\text{f}} \eq v_{\text{i}} + a_{\text{ave}} \spc \parens{t_{\text{f}} - t_{\text{i}}}
$

<br>

**Step 4: Plug In Values:**

<br>

$
\begin{array}{l l}
v_{\text{f}} & \eq v_{\text{i}} + a_{\text{ave}} \spc \parens{t_{\text{f}} - t_{\text{i}}} \\&\\
& \eq 24 \mps + 3 \mpss \spc \parens{13 \seconds - 5 \seconds} \\&\\
& \eq 24 \mps + 3 \mpss \spc \parens{8 \seconds} \\&\\
& \eq 24 \mps + 24 \mps \\&\\
& \eq 48 \mps
\end{array}
$

<br>

**Answer:**

<br>

The car will reach a velocity of **(48) meters per second** in the positive direction at (13) seconds.

t_{\text{f}} & \eq t_{\text{i}} + \dsfrac{v_{\text{f}} - v_{\text{i}}}{a_{\text{ave}}} \\&\\
& \eq 4 \seconds + \dsfrac{30 \mps - 8 \mps}{2 \mpss} \\&\\
& \eq 4 \seconds + \dsfrac{22 \mps}{2 \mpss} \\&\\
& \eq 4 \seconds + 11 \seconds \\&\\
& \eq 15 \seconds
\end{array}
$

<br>

**Answer:**

<br>

The bicycle will reach a velocity of **30 meters per second** in the positive direction at 15 seconds.

###<font size = 4> <b>Example 4:</b></font>

A runner is traveling along a straight track. At time 3 $\seconds$, they pass the 10 m mark while moving at 6 $\mps$ in the positive direction. The runner has a constant acceleration of 2 $\mpss$. At what time will the runner reach a position of 100 m?

<br>

**Step 1: Get the Values from the Problem:**

<br>

$
\begin{array}{l l l}
\textrm{Positions:} & x_{\text{i}} = 10 \textrm{ m}\textrm{, } &
x_{\text{f}} = 100 \textrm{ m} \\&&\\
\textrm{Times:} & t_{\text{i}} = 3 \seconds \textrm{, } & t_{\text{f}} = \textrm{?} \\&&\\
\textrm{Velocities:} & v_{\text{i}} = 6 \mps\textrm{, } & \\&&\\
\textrm{Acceleration:} & a = 2 \mpss & \\&&\\
\end{array}
$

<br>

**Step 2: Get the Relevant Equation(s):**

<br>

Since the final velocity is unknown, use the position equation for constant acceleration:

<br>

$
x_{\text{f}} \eq x_{\text{i}} + v_{\text{i}} \spc \parens{t_{\text{f}} - t_{\text{i}}} + \frac{1}{2} \spc a \spc \parens{t_{\text{f}} - t_{\text{i}}}^2
$

<br>

**Step 3: Evaluate the Expression:**

<br>

Let $\Delta t = t_{\text{f}} - t_{\text{i}}$

<br>

$
x_{\text{f}} - x_{\text{i}} \eq v_{\text{i}} \spc \Delta t + \frac{1}{2} \spc a \spc \parens{\Delta t}^2
$

<br>

Rearrange into a quadratic equation:

<br>

$
\frac{1}{2} \spc a \spc \Delta t^2 + v_{\text{i}} \spc \Delta t - \parens{x_{\text{f}} - x_{\text{i}}} \eq 0
$

<br>

Quadratic Formula: $A \spc \Delta t^2 + B \spc \Delta t + C = 0$

<br>

Thus:

<br>

$
\begin{array}{l l}
A = & \frac{1}{2} \spc a \\&\\
B = & v_{\text{i}} \\&\\
C = & \parens{x_{\text{f}} - x_{\text{i}}}
\end{array}
$

<br>


Using the quadratic formula:

<br>

$
\Delta t \eq
\dsfrac{-v_{\text{i}} \pm \sqrt{v_{\text{i}}^2 - 2 \spc a \parens{-90}}}{a}
$

<br>

$
\Delta t \eq
\frac{-6 \pm \sqrt{396}}{2}
$

<br>

**Step 3: Plug In Values:**

<br>

$
\begin{array}{l l}
0 & \eq \frac{1}{2}\parens{2 \mpss}\Delta t^2

* 6 \mps\Delta t

- \parens{100 \textrm{ m} - 10 \textrm{ m}} \&&\
  0 & \eq \Delta t^2 + 6\Delta t - 90
  \end{array}
  $



<br>

The physically meaningful solution is:

$
\Delta t \approx 6.95 \seconds
$

Therefore,

$
t_{\text{f}} \eq t_{\text{i}} + \Delta t
$

$
t_{\text{f}} \eq 3 \seconds + 6.95 \seconds
$

$
t_{\text{f}} \eq 9.95 \seconds
$

<br>

**Answer:**

<br>

The runner will reach a position of **100 meters** at approximately **9.95 seconds**.

##<font size = 5> <b>6.1 Instantaneous Acceleration Formula(s)</b> </font>

##<font size = 5> <b>6.2 Instantaneous Acceleration Example Problem(s)</b> </font>

###<font size = 4> <b>Example 1:</b></font>

The velocity of a particle moving along the x-axis is described by:

<br>

$$v\of{t} = 4t^2 - 3t + 2$$

<br>

where v is measured in meters per second and t is measured in seconds. Determine the instantaneous acceleration of the particle at t = 3.0 s.

<br>

**Step 1: Get the Values from the Problem:**

<br>

$
\begin{array}{l l l}
\textrm{Time:} & t = 3.0 \textrm{ s}
\end{array}
$

<br>

**Step 2: Get the Relevant Equation(s):**

<br>

$
a\of{t} \eq \dsfrac{d \spc v\of{t}}{d \spc t}
$

<br>

**Step 3: Evaluate the Expression:**

<br>

$
\begin{array}{l l}
v\of{t} & \eq \dsfrac{d \spc x\of{t}}{d \spc t} \\&\\
& \eq \dsfrac{d}{d \spc t} \spc \parens{4t^2 - 3t + 2} \\&\\
& \eq \dsfrac{d}{d \spc t} \spc \parens{4t^2} - \dsfrac{d}{d \spc t} \spc \parens{3t} + \dsfrac{d}{d \spc t} \spc \parens{2} \\&\\
& \eq 8t - 3 + 0 \\&\\
& \eq 8t - 3
\end{array}
$

<br>

**Step 4: Plug In Values:**

<br>

$
\begin{array}{l l}
a\of{t} & \eq 8 \frac{\textrm{m}}{\textrm{s}^3} \spc \parens{3.0 \textrm{ s}} - 3 \mpss \\&\\
& \eq 24 \mpss - 3 \mpss \\&\\
& \eq 21 \mpss
\end{array}
$
