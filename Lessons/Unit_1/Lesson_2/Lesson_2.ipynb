##<font size = 5> <b>1.1 Average Velocity Formula(s)</b> </font>

1. $\hspace{1 pc}
v_{\text{ave}} \eq
\dsfrac{x_{\text{f}} - x_{\text{i}}}{t_{\text{f}} - t_{\text{i}}}
$

<br>

---

<br>

##<font size = 5> <b>1.2 Average Velocity Example Problem(s)</b> </font>

###<font size = 4> <b>Example 1:</b></font>

A runner passes a marker located at 25 m along a straight track at t = 5s. At t = 17 s, the runner passes a marker located at 100 m along the straight track. What is the average velocity of the runner during this time interval?

<br>

**Step 1: Get the Values from the Problem:**

<br>

$
\begin{array}{l l l}
\textrm{Positions:} & x_{\text{i}} = 25 \textrm{ m} \textrm{, } &
x_{\text{f}} = 100 \textrm{ m} \\&&\\
\textrm{Times:} & t_{\text{i}} = 5  \seconds \textrm{, } & t_{\text{f}} = 17 \seconds
\end{array}
$

<br>

**Step 2: Get the Relevant Equation(s):**

<br>

$
v_{\text{ave}} \eq
\dsfrac{x_{\text{f}} - x_{\text{i}}}{t_{\text{f}} - t_{\text{i}}}
$

<br>

**Step 3: Plug In Values:**

<br>

$
\begin{array}{l l}
v_{\text{ave}}
& \eq \dsfrac{100 \textrm{ m} - 25 \textrm{ m}}{17 \seconds - 5 \seconds} \\&\\
& \eq \dsfrac{75 \textrm{ m}}{12 \seconds} \\&\\
& \eq 6.25 \spc \mps
\end{array}
$

<br>

##<font size = 5> <b>2.2 Motion with Constant Velocity Example Problem(s)</b> </font>

###<font size = 4> <b>Example 1:</b></font>

A train moves along a straight section of track with a constant velocity of 15 m/s in the positive direction. At t = 10 s, the train is 320 m from a station. How far from the station will the train be at t = 25 s?

<br>

**Step 1: Get the Values from the Problem:**

<br>

$
\begin{array}{l l l}
\textrm{Positions:} & x_{\text{i}} = 320 \textrm{ m} \textrm{, } &
x_{\text{f}} = ??? \textrm{ m} \\&&\\
\textrm{Times:} & t_{\text{i}} = 10  \seconds \textrm{, } & t_{\text{f}} = 25 \seconds \\&\\
v_{\textrm{ave}} = 15 \mps
\end{array}
$

<br>

**Step 2: Get the Relevant Equation(s):**

<br>

$
x_{\text{f}} \eq
x_{\text{f}} \eq x_{\text{i}} + v_{\text{ave}} \spc \parens{t_{\text{f}} - t_{\text{i}}}
$

<br>

**Step 3: Plug In Values:**

<br>

$
\begin{array}{l l}
x_{\text{f}} & \eq 320 \textrm{ m} + 15 \mps \spc \parens{25 \textrm{ s} - \textrm{10 s}} \\&\\
& \eq 320 \textrm{ m} + 15 \mps \spc \parens{15 \textrm{ s}} \\&\\
& \eq 320 \textrm{ m} + 225 \textrm{ m} \\&\\
& \eq 545 \textrm{ m}
\end{array}

#<font size = 6> <b> 3. Instantaneous Velocity </b> </font>

---

##<font size = 5> <b>3.1 Instantaneous Velocity Formula(s)</b> </font>

1. $\hspace{1 pc}
v\of{t} \eq \dsfrac{d \spc x\of{t}}{d \spc t}$

<br>

---

<br>

##<font size = 5> <b>3.2 Instantaneous Velocity Example Problem(s)</b> </font>

###<font size = 4> <b>Example 1:</b></font>

The position of a particle moving along the x-axis is described by:

<br>

$$x\of{t} = 3t^2 + 2t - 5$$

<br>

where x is measured in meters and t is measured in seconds. Determine the instantaneous velocity of the particle at t = 4.0 s.

<br>

**Step 1: Get the Values from the Problem:**

<br>

$
\begin{array}{l l l}
\textrm{Time:} & t = 4.0 \textrm{ s}
\end{array}
$

<br>

**Step 2: Get the Relevant Equation(s):**

<br>

$
v\of{t} \eq \dsfrac{d \spc x\of{t}}{d \spc t}
$

<br>

**Step 3: Evaluate the Expression:**

<br>

$
\begin{array}{l l}
v\of{t} & \eq \dsfrac{d \spc x\of{t}}{d \spc t} \\&\\
& \eq \dsfrac{d}{d \spc t} \spc \parens{3t^2 + 2t - 5} \\&\\
& \eq \dsfrac{d}{d \spc t} \spc \parens{3t^2} + \dsfrac{d}{d \spc t} \spc \parens{2t} - \dsfrac{d}{d \spc t} \spc \parens{5} \\&\\
& \eq 6t + 2 + 0 \\&\\
& \eq 6t + 2
\end{array}
$

<br>

**Step 4: Plug In Values:**

<br>

$
\begin{array}{l l}
v\of{t} & \eq 6 \mpss \spc \parens{4.0 \textrm{ s}} + 2 \mps \\&\\
& \eq 24 \mps + 2 \mps \\&\\
& \eq 26 \mps
\end{array}
$

<br>

$
\begin{array}{l l l}
\textrm{Velocities:} & v_{\text{i}} = 24 \mps\textrm{, } &

<br>

$

$

<br>

**Step 4: Plug In Values:**

<br>

$
\begin{array}{l l}
v_{\text{f}} & \eq v_{\text{i}} + a_{\text{ave}} \spc \parens{t_{\text{f}} - t_{\text{i}}} \\&\\

#<font size = 6> <b> 6. Instantaneous Acceleration </b> </font>

---

##<font size = 5> <b>6.1 Instantaneous Acceleration Formula(s)</b> </font>

1. $\hspace{1 pc}
a\of{t} \eq \dsfrac{d \spc v\of{t}}{d \spc t}$

2. $\hspace{1 pc}
a\of{t} \eq \dsfrac{d^2 \spc x\of{t}}{d \spc t^2}$

<br>

---

<br>

##<font size = 5> <b>6.2 Instantaneous Acceleration Example Problem(s)</b> </font>

###<font size = 4> <b>Example 1:</b></font>

The velocity of a particle moving along the x-axis is described by:

<br>

$$v\of{t} = 4t^2 - 3t + 2$$

<br>

where v is measured in meters per second and t is measured in seconds. Determine the instantaneous acceleration of the particle at t = 3.0 s.

<br>

**Step 1: Get the Values from the Problem:**

<br>

$
\begin{array}{l l l}
\textrm{Time:} & t = 3.0 \textrm{ s}
\end{array}
$

<br>

**Step 2: Get the Relevant Equation(s):**

<br>

$
a\of{t} \eq \dsfrac{d \spc v\of{t}}{d \spc t}
$

<br>

**Step 3: Evaluate the Expression:**

<br>

$
\begin{array}{l l}
v\of{t} & \eq \dsfrac{d \spc x\of{t}}{d \spc t} \\&\\
& \eq \dsfrac{d}{d \spc t} \spc \parens{4t^2 - 3t + 2} \\&\\
& \eq \dsfrac{d}{d \spc t} \spc \parens{4t^2} - \dsfrac{d}{d \spc t} \spc \parens{3t} + \dsfrac{d}{d \spc t} \spc \parens{2} \\&\\
& \eq 8t - 3 + 0 \\&\\
& \eq 8t - 3
\end{array}
$

<br>

**Step 4: Plug In Values:**

<br>

$
\begin{array}{l l}
a\of{t} & \eq 8 \frac{\textrm{m}}{\textrm{s}^3} \spc \parens{3.0 \textrm{ s}} - 3 \mpss \\&\\
& \eq 24 \mpss - 3 \mpss \\&\\
& \eq 21 \mpss
\end{array}
$

```python
##================================================================================================##
## Import Libraries:
##================================================================================================##

import json
import textwrap

from google.colab import ai


##================================================================================================##
## Function: generate_problem
##================================================================================================##

def generate_problem(topic, variables):
    """
    Generate a physics word problem using the Google Colab AI model.

    Inputs:
        topic     - String describing the physics topic.
        variables - List of strings describing the variables and their meanings.

    Outputs:
        problem   - Problem text with variable placeholders.
        values    - Dictionary containing values for each variable.
    """

    #----------------------------------------------------------------------------------------------#
    # Format Variable Descriptions:
    #----------------------------------------------------------------------------------------------#

    variable_text = "\n".join(
        f"- {{{variable['name']}}}: {variable['description']}"
        for variable in variables
    )

    #----------------------------------------------------------------------------------------------#
    # Generate Example JSON:
    #----------------------------------------------------------------------------------------------#

    example_values = ",\n".join(
        f'"{variable["name"]}": {variable["example"]}'
        for variable in variables
    )

    example_problem = " ".join(
        f"{{{variable['name']}}}"
        for variable in variables
    )

    example = f"""
{{
    "Problem": "Example problem using {example_problem}.",
    "Values": {{
        {example_values}
    }}
}}
"""

    #----------------------------------------------------------------------------------------------#
    # Construct LLM Prompt:
    #----------------------------------------------------------------------------------------------#

    llm_prompt = f"""
Output Format: JSON format. Entry 1 is a word problem. Entry 2 is the variable values.

Topic:
{topic}

Variables:
{variable_text}

Entry 1:
Generate an interesting physics word problem for a {topic} problem.

The problem must use the variables listed above and represent those variables using
their names inside curly braces. For example, use {{ti}} rather than a numerical
value for initial time.

Entry 2:
Provide a physically reasonable numerical value for every variable.

Output only the JSON and no additional commentary or questions.

Example Output:

{example}
"""

    #----------------------------------------------------------------------------------------------#
    # Generate Problem:
    #----------------------------------------------------------------------------------------------#

    llm_response = ai.generate_text(llm_prompt)

    #----------------------------------------------------------------------------------------------#
    # Convert JSON Response to Dictionary:
    #----------------------------------------------------------------------------------------------#

    problem_dict = json.loads(llm_response)

    problem = problem_dict["Problem"]
    values = problem_dict["Values"]

    #----------------------------------------------------------------------------------------------#
    # Ensure Time Variables are Ordered:
    #----------------------------------------------------------------------------------------------#

    if "ti" in values and "tf" in values:

        if values["ti"] == values["tf"]:
            values["tf"] = values["ti"] * 1.1

        if values["ti"] > values["tf"]:
            values["ti"], values["tf"] = values["tf"], values["ti"]

    #----------------------------------------------------------------------------------------------#
    # Insert Variable Values into Problem:
    #----------------------------------------------------------------------------------------------#

    problem = problem.format(**values)

    return problem, values


##================================================================================================##
## Function: solve_problem
##================================================================================================##

def solve_problem(topic, values):
    """
    Solve a physics problem based on its topic and variable values.

    Inputs:
        topic  - String describing the physics topic.
        values - Dictionary containing the problem variables.

    Outputs:
        solution_value - Numerical value of the solution.
        solution_units - List of acceptable units.
    """

    #----------------------------------------------------------------------------------------------#
    # 1D Average Velocity:
    #----------------------------------------------------------------------------------------------#

    if topic.lower() == "1d average velocity":

        ti = values["ti"]
        tf = values["tf"]
        xi = values["xi"]
        xf = values["xf"]

        solution_value = (xf - xi) / (tf - ti)

        solution_units = [
            "m/s",
            "meters per second",
            "mps"
        ]

    #----------------------------------------------------------------------------------------------#
    # Unknown Topic:
    #----------------------------------------------------------------------------------------------#

    else:
        raise ValueError(
            f"No solution method has been defined for topic: {topic}"
        )

    return solution_value, solution_units


##================================================================================================##
## Define Problem:
##================================================================================================##

topic = "1d average velocity"

variables = [
    {
        "name": "ti",
        "description": "initial time (in seconds)",
        "example": 2.5
    },
    {
        "name": "tf",
        "description": "final time (in seconds)",
        "example": 20.8
    },
    {
        "name": "xi",
        "description": "initial position (in meters)",
        "example": 15.3
    },
    {
        "name": "xf",
        "description": "final position (in meters)",
        "example": 1063.7
    }
]


##================================================================================================##
## Generate Problem:
##================================================================================================##

problem, values = generate_problem(
    topic,
    variables
)


##================================================================================================##
## Solve Problem:
##================================================================================================##

solution_value, solution_units = solve_problem(
    topic,
    values
)


##================================================================================================##
## Print Problem:
##================================================================================================##

wrapped_problem = textwrap.fill(
    problem,
    width=100
)

print(wrapped_problem)

print()
print(f"Solution: {solution_value:.3f} {solution_units[0]}")
```
